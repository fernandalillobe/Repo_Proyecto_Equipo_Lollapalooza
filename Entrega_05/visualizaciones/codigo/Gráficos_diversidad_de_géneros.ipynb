### **5. Ordenar datos**

```python
lolla_generos = lolla.melt(
    id_vars=["edicion lolla "],
    value_vars=["genero musical 01", "genero musical 02", "genero musical 03"],
    value_name="genero"
).dropna()

lolla_generos
```

```python
lolla_generos = lolla_generos.drop(columns=["variable"])
lolla_generos
```

```python
tabla_generos = lolla_generos.groupby(["edicion lolla ", "genero"]).size(). reset_index(name="cantidad")
tabla_generos
```

```python
top_generos = tabla_generos.groupby("genero")["cantidad"].sum().nlargest(10).index
tabla_top = tabla_generos[tabla_generos["genero"].isin(top_generos)]
tabla_top
```

### 6. **Gráfico con Altair**

```python
grafico = alt.Chart(tabla_top).mark_bar().encode(
    x="edicion lolla :O",
    y="cantidad:Q",
    color="genero:N",
    tooltip=["edicion lolla ", "genero", "cantidad"]
).properties(
    title="Diversificación de géneros musicales por edición"
)

grafico
```

```python
lolla_generos_limpio = lolla_generos.dropna(subset=["genero"])
lolla_generos_limpio
```

```python
top15_generos = lolla_generos_limpio.groupby("genero")["genero"].count().nlargest(15).index
tabla_top15 = lolla_generos_limpio[lolla_generos_limpio["genero"].isin(top15_generos)]
tabla_top15 = tabla_top15.groupby(["edicion lolla ", "genero"]).size().reset_index(name="cantidad")
tabla_top15
```

```python
grafico_norm = alt.Chart(tabla_top15).mark_bar().encode(
    y="edicion lolla :O",
    x=alt.X("cantidad:Q", stack="normalize", axis=alt.Axis(format="%")),
    color="genero:N",
    tooltip=["edicion lolla ", "genero", "cantidad"]
).properties(
    title="Diversificación relativa de géneros musicales por edición"
)

grafico_norm
```

```python
total_bandas = lolla.groupby("edicion lolla ").size().reset_index(name="total_bandas")
total_bandas
```

```python
total_generos = lolla_generos_limpio.groupby("edicion lolla ")["genero"].nunique().reset_index(name="total_generos")
total_generos
```

```python
tabla_combinada = total_bandas.merge(total_generos, on="edicion lolla ")
tabla_combinada
```

```python
barras = alt.Chart(tabla_combinada).mark_bar(color="steelblue", opacity=0.7).encode(
    x="edicion lolla :O",
    y=alt.Y("total_bandas:Q", title="Total de bandas"),
    tooltip=["edicion lolla ", "total_bandas", "total_generos"]
)

linea = alt.Chart(tabla_combinada).mark_line(color="red", point=True).encode(
    x="edicion lolla :O",
    y=alt.Y("total_generos:Q", title="Total de géneros únicos"),
    tooltip=["edicion lolla ", "total_generos"]
)

grafico_combinado = (barras + linea).properties(
    title="Crecimiento de bandas y diversidad de géneros por edición"
).resolve_scale(y="independent")

grafico_combinado
```

```python
import unicodedata

def normaliza(s):
    s = s.lower().strip()
    s = ''.join(c for c in unicodedata.normalize('NFD', s) if unicodedata.category(c) != 'Mn')
    return s

def categoria(g):
    n = normaliza(g)
    if 'rock' in n or 'punk' in n or 'grunge' in n or 'metal' in n or 'emo' in n:
        return 'Rock'
    if 'pop' in n and 'hip' not in n:
        return 'Pop'
    if 'electro' in n or 'house' in n or 'techno' in n or 'tecno' in n or 'trance' in n or 'dance' in n or 'rave' in n or 'dubstep' in n or 'drum' in n or 'jungle' in n or 'breakbeat' in n or 'bass' in n:
        return 'Electrónica'
    if 'hip hop' in n or 'rap' in n or 'trap' in n or 'drill' in n:
        return 'Hip Hop / Rap'
    if 'reggaeton' in n or 'regueton' in n or 'reggueton' in n or 'urbano' in n or 'urbana' in n or 'rkt' in n or 'dembow' in n:
        return 'Urbano / Reguetón'
    if 'r&b' in n or 'soul' in n:
        return 'R&B / Soul'
    if 'reggae' in n or 'ska' in n or 'dub' in n:
        return 'Reggae / Ska'
    if 'folk' in n or 'folclor' in n or 'andin' in n or 'cueca' in n or 'trova' in n:
        return 'Folk / Folclor'
    if 'cumbia' in n or 'salsa' in n or 'tropical' in n:
        return 'Latino tropical'
    if 'jazz' in n:
        return 'Jazz'
    return 'Otros'
```

```python
lolla_generos_limpio["categoria"] = lolla_generos_limpio["genero"].apply(categoria)

destacados = ["Rock", "Electrónica", "Pop"]
lolla_generos_limpio["destacado"] = lolla_generos_limpio["categoria"].apply(
    lambda x: x if x in destacados else "Otros"
)

tabla_destacada = lolla_generos_limpio.groupby(["edicion lolla ", "destacado"]).size().reset_index(name="cantidad")
tabla_destacada
```

```python
alt.renderers.enable('mimetype')
```

```python
grafico_destacado = alt.Chart(tabla_destacada).mark_bar().encode(
    y="edicion lolla :O",
    x=alt.X("cantidad:Q", stack="normalize", axis=alt.Axis(format="%")),
    color=alt.Color("destacado:N", scale=alt.Scale(
        domain=["Rock", "Electrónica", "Pop", "Otros"],
        range=["#e63946", "#457b9d", "#f4a261", "#cccccc"]
    )),
    tooltip=["edicion lolla ", "destacado", "cantidad"]
).properties(
    title="Géneros predominantes por edición: Rock, Electrónica y Pop"
)

grafico_destacado
```
