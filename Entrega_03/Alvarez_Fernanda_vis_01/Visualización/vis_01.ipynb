```python
lolla.columns = lolla.columns.str.strip()
lolla['headliner'] = lolla['headliner'].str.strip()
lolla['nacionalidad'] = lolla['nacionalidad'].str.strip()
```

```python
lolla_headliners = lolla[lolla['headliner'] == 'Sí'].copy()
```

```python
alt.Chart(lolla_headliners).mark_bar().encode(
    x=alt.X('count():Q', title='Cantidad de Headliners'),
    y=alt.Y('nacionalidad:N', sort='-x', title='País'),
    color=alt.Color('nacionalidad:N', legend=None),
    tooltip=['nacionalidad', 'count()']
).properties(
    title='Diversidad de Nacionalidades en Headliners',
    width=500
)
```
