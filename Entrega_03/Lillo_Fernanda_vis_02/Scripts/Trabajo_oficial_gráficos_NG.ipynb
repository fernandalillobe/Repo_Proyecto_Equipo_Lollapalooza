```python
lolla.columns = lolla.columns.str.strip()
```

```python
artistas_chilenos = lolla[lolla["nacionalidad"].str.contains("chile", case=False, na=False)].copy()
```

```python
conteo_chilenos = artistas_chilenos.groupby("edicion lolla").size().reset_index(name="cantidad")
```

```python
grafico = alt.Chart(conteo_chilenos).mark_bar(color="teal").encode(
    x=alt.X("edicion lolla:O", title="Año de la edicion"),
    y=alt.Y("cantidad:Q", title="Número de artistas chilenos"),
    tooltip=[
        alt.Tooltip("edicion lolla", title="Edición"),
        alt.Tooltip("cantidad", title="Total Artistas")
    ]
).properties(
    title="Representación de artistas chilenos en Lollapalooza por año",
    width=600,
    height=400
).interactive()
grafico
grafico.save("representacion_chilena.html")
```

```python
import pandas as pd
import altair as alt
from PIL import Image
lolla.colums = lolla.columns.str.strip()
artistas_chilenos = lolla[lolla['nacionalidad'].str.contains('chile', case=False, na=False)].copy()
conteo_chilenos = artistas_chilenos.groupby('edicion lolla').size().reset_index(name='cantidad')
grafico = alt.Chart(conteo_chilenos).mark_bar(color='teal').encode(
    x=alt.X('edicion lolla:O', title='Año de la Edición'),
    y=alt.Y('cantidad:Q', title='Número de Artistas Chilenos'),
    tooltip=['edicion lolla', 'cantidad']
).properties(
    title='Representación de Artistas Chilenos en Lollapalooza',
    width=600,
    height=400
)
grafico.save('temp_grafico.png', scale_factor=2.0)
img_png = Image.open('temp_grafico.png')
img_rgb = img_png.convert('RGB')
img_rgb.save('grafico_lollapalooza.jpg', quality=95)
print("¡Archivo grafico_lollapalooza.jpg creado con éxito!")
```
