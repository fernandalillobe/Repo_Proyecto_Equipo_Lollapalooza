### **5. Ordenar datos**

```python
lolla_generos = lolla.melt(
    id_vars=["edicion lolla "],
    value_vars=["genero musical 01", "genero musical 02", "genero musical 03"],
    value_name="genero"
).dropna()

lolla_generos
```

```python
lolla_generos = lolla_generos.drop(columns=["variable"])
lolla_generos
```

```python
tabla_generos = lolla_generos.groupby(["edicion lolla ", "genero"]).size(). reset_index(name="cantidad")
tabla_generos
```

```python
top_generos = tabla_generos.groupby("genero")["cantidad"].sum().nlargest(10).index
tabla_top = tabla_generos[tabla_generos["genero"].isin(top_generos)]
tabla_top
```

### 6. **Gráfico con Altair**

```python
grafico = alt.Chart(tabla_top).mark_bar().encode(
    x="edicion lolla :O",
    y="cantidad:Q",
    color="genero:N",
    tooltip=["edicion lolla ", "genero", "cantidad"]
).properties(
    title="Diversificación de géneros musicales por edición"
)

grafico
```
